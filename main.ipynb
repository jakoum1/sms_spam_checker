```python
import pandas as pd
```

```python
to_drop=['Unnamed: 2','Unnamed: 2','Unnamed: 3','Unnamed: 4']
df=df.drop(columns=to_drop)
```

```python
#lower case the messagese
df['v2']=df['v2'].str.lower()
df['v2']
```

```python
#remove punctuation marks
import re 
df['v2']=df['v2'].apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

```python
df['v2']
```

```python
#apply tokenisation and remove stop words
import nltk 
nltk.download("stopwords")
nltk.download("punkt")
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words=stopwords.words("english")
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
df['v2']=df['v2'].apply(word_tokenize)
```

```python
df
```

```python
def remove_stop_words(tokens) :
    return [word for word in tokens if word not in stop_words]
```

```python
df.apply(remove_stop_words)
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download("wordnet")
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
def lem(tokens) :
    return [lemmatizer.lemmatize(word) for word in tokens]
```

```python
df['v2'].apply(lem)
```
